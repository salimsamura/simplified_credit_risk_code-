```python
from sklearn.model_selection import cross_val_predict

validation_pd = cross_val_predict(
    xgb_model,
    x_train,
    y_train,
    cv=cv,
    method='predict_proba'
)[:, 1]
```

```python
validation_pd
```

```python
results = []
actual = y_train.to_numpy()

for cutoff in [0.10, 0.15, 0.20, 0.25, 0.30]:
    approve = validation_pd <= cutoff

    results.append({
        'PD cutoff': cutoff,
        'Approval rate': approve.mean(),
        'Default rate among approved':
            actual[approve].mean() if approve.any() else np.nan,
        'Default recall':
            (~approve)[actual == 1].mean()
    })

pd.DataFrame(results).style.format('{:.1%}')
```


*based on this table at a cutoff rate of 15% or less if 100 customers apply 42 people got approved and 58 got rejected
out of the 42*0.111= 4 defaulted 

*on the otherhand if we choose a cutoff of 30% or less if 100 customers apply 100 * 0.874= 87.4 will be approved and 100-87.4 = 12 rejected and out of the 87.4 * 0.163= 14 will default leaving a total of 73 will fully pay out their loan

```python
# highest probability of bad will accept is 30% meanning lowest approval rate of good is 70%

probability_target_table['decision'] = np.where(probability_target_table['Probability_of_default']<=0.30,'Approve', 'Reject')
probability_target_table.head(100)
```

* GINI AND KOLMOGOROV_SMIRNOV TEST

```python
probability_target_table
```

```python
probability_target_table=probability_target_table.sort_values('Probability_of_default', ascending=False)
probability_target_table.head()
```

```python
probability_target_table=probability_target_table.reset_index()
probability_target_table.head()
```

```python
probability_target_table['Cumulative N Population'] = probability_target_table.index + 1
probability_target_table['Cumulative N Bad'] =probability_target_table['y_test'].cumsum()
probability_target_table['Cumulative N Good'] = probability_target_table['Cumulative N Population'] - probability_target_table['y_test'].cumsum() 
probability_target_table.head()
```

```python
probability_target_table['cumulative % population'] = (
    probability_target_table['Cumulative N Population']
    / probability_target_table.shape[0])

probability_target_table['cumulative % Good'] = (
    probability_target_table['Cumulative N Good']
    / (probability_target_table.shape[0] - probability_target_table['y_test'].sum()))

probability_target_table['cumulative % Bad'] = (
    probability_target_table['Cumulative N Bad']
    / probability_target_table['y_test'].sum())

probability_target_table.head()
```

```python
plt.plot(probability_target_table['cumulative % population'], probability_target_table['cumulative % Bad'])
plt.plot(probability_target_table['cumulative % population'], probability_target_table['cumulative % population'], linestyle='--', color='k')
plt.xlabel('cumulative % population')
plt.ylabel('cumulative % Bad')
plt.title('GINI')
plt.show()
```

* K-S

```python
probability_target_table
```

```python
plt.plot(probability_target_table.Probability_of_default, probability_target_table['cumulative % Bad'], color='r', label='Bad')
plt.plot(probability_target_table.Probability_of_default, probability_target_table['cumulative % Good'], color='b', label='Good')
plt.xlabel('Estimated Probability of Default')
plt.ylabel('Cumulative % population')
plt.title('Kolmogorov-Smirnov (KS) Curve')
plt.legend()
plt.show()
```

```python
ks= max(probability_target_table['cumulative % Bad'] - probability_target_table['cumulative % Good'])
print(f'KS: {ks:.3f} ({ks:.1%})')
```

* At the point of greatest separation, the model flags a share of actual defaulters that is 23.2 percentage points higher than the share of repaid borrowers
